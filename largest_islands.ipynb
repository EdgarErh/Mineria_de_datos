```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('largest-islands.csv')
df.head(10)
```

### Project Ideas

- What are the 10 largest islands in the tropics?

- What are the largest islands in each `region`?

- Create a line graph with `area` on the y-axis and `rank` on the x-axis. The data should be ordered by `rank`, from largest to smallest.

- What islands are composed of multiple countries?
	- Hint: try `countries.str.contains(",")`

```python
# YOUR CODE HERE (add more cells as needed)
df_10 = df.nlargest(10, 'area')
df_10 = df_10[['island', 'area']]
df_10
```

```python

df_ordenado = df.sort_values(by='area', ascending=False)

df_islas_max_region = df_ordenado.groupby('region').head(1)

print(df_islas_max_region[['region', 'island', 'area']])
```

```python
df_sorted = df.sort_values(by='rank', ascending=False)

plt.figure(figsize=(10, 6))
plt.plot(df_sorted['rank'], df_sorted['area'], marker='o', linestyle='-', color='green')

plt.xlabel('Rank')
plt.ylabel('Area')
plt.title('Área vs Rank (Ordenado de Rank mayor a menor)')

plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
